```python
import pandas as pd
import geopandas as gpd

# 1. Load the GTFS stops file
stops_path = r"C:/Users/mpanta/Documents/AT3_KGG375/METRO Data/stops.txt"
stops_df = pd.read_csv(stops_path)

# 2. Convert to a spatial GeoDataFrame (WGS84 coordinate system)
gdf_stops = gpd.GeoDataFrame(
    stops_df, 
    geometry=gpd.points_from_xy(stops_df['stop_lon'], stops_df['stop_lat']),
    crs="EPSG:4326"
)

# 3. Filter spatially for Greater Burnie bounding box
# [longitude min:max, latitude min:max]
burnie_stops = gdf_stops.cx[145.80:146.00, -41.15:-41.00].copy()

# 4. Save to your project folder
output_path = r"C:/Users/mpanta/Documents/AT3_KGG375/burnie_bus_stops.geojson"
burnie_stops.to_file(output_path, driver="GeoJSON")

print(f"Extracted {len(burnie_stops)} bus stops for Burnie!")
```
