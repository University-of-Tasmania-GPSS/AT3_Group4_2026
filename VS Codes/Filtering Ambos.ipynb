```python
import geopandas as gpd

# 1. Load the statewide ambulance stations shapefile
ambulance_path = r"C:/Users/mpanta/Documents/AT3_KGG375/Ambulance Stations TAS/dhhs_ambulance_stations_statewide.shp"
gdf_amb = gpd.read_file(ambulance_path)

# 2. Filter for Burnie, Wynyard, and Ulverstone
stations_to_keep = ['Burnie', 'Wynyard', 'Ulverstone']
pattern = '|'.join(stations_to_keep)

filtered_amb = gdf_amb[
    gdf_amb.apply(lambda row: row.astype(str).str.contains(pattern, case=False).any(), axis=1)
].copy()

# 3. Save the filtered points to your project folder
output_path = r"C:/Users/mpanta/Documents/AT3_KGG375/north_west_ambulances.geojson"
filtered_amb.to_file(output_path, driver="GeoJSON")

print(f"Successfully saved {len(filtered_amb)} stations!")
```
