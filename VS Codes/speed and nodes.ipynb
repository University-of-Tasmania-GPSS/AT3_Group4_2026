```python
import osmnx as ox

# Download drivable street network for Burnie, Tasmania with speed attributes
G = ox.graph_from_place(
    'Burnie, Tasmania, Australia', network_type='drive'
)

# Add imputed maximum speeds where tags might be missing
G = ox.add_edge_speeds(G)

# Convert graph to GeoDataFrames and save as a GeoPackage
nodes, edges = ox.graph_to_gdfs(G)
edges.to_file('burnie_network.gpkg', layer='edges', driver='GPKG')
nodes.to_file('burnie_network.gpkg', layer='nodes', driver='GPKG')

import os

# Define your target folder and file path
output_dir = r"C:\Users\mpanta\Documents\AT3_KGG375"  # Replace with your actual folder path
os.makedirs(output_dir, exist_ok=True)
gpkg_path = os.path.join(output_dir, "burnie_network.gpkg")

# Convert graph to GeoDataFrames and save both edges and nodes into the GeoPackage
nodes, edges = ox.graph_to_gdfs(G)
edges.to_file(gpkg_path, layer="edges", driver="GPKG")
nodes.to_file(gpkg_path, layer="nodes", driver="GPKG")

print(f"Saved successfully to: {gpkg_path}")
```
