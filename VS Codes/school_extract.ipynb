```python
import osmnx as ox
import geopandas as gpd

# Download all schools in Burnie
schools_burnie = ox.features_from_place("Burnie, Tasmania, Australia", tags={"amenity": "school"})

# Convert polygons to point locations
schools_burnie["geometry"] = schools_burnie.geometry.centroid

# Save to file
output_path = r"C:/Users/mpanta/Documents/AT3_KGG375/all_burnie_schools.geojson"
schools_burnie.to_file(output_path, driver="GeoJSON")
print(f"Extracted {len(schools_burnie)} schools in Burnie.")
```
